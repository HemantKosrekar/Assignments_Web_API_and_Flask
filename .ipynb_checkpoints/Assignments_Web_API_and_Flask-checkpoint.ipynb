### 1. What is a Web API?
--> A Web API is an Application Programming Interface that works over the web.  
--> It allows communication between two software systems using the internet.  
--> It helps clients (mobile apps, websites) interact with server data or services.  
--> It uses standard protocols like HTTP/HTTPS for communication.  
--> It commonly exchanges data in formats like JSON or XML.  

### 2. How does a Web API differ from a web service?
--> A Web API is more flexible and can be used by any client (browser, mobile app, etc.), while a traditional web service usually follows stricter rules.  
--> Web APIs mainly use lightweight formats like JSON, but web services often rely on XML.  
--> Web APIs are simpler and faster to implement, whereas web services can be more complex because they follow specific standards like SOAP.  
--> Web APIs work smoothly with modern web applications, while web services are mostly used in older or enterprise-level systems.
